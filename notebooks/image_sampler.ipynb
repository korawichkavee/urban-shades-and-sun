```python
import os
import pandas as pd
import random
import shutil
random.seed(42)
```

```python
#from chatGPT
def copy_random_files(source_dir, dest_dir, num_files=7):
    """
    Randomly selects a specified number of files from a source directory
    and copies them to a destination directory.

    Args:
        source_dir (str): The path to the source directory.
        dest_dir (str): The path to the destination directory.
        num_files (int): The number of files to randomly select and copy.
    """
    if not os.path.exists(source_dir):
        print(f"Error: Source directory '{source_dir}' does not exist.")
        return

    if not os.path.exists(dest_dir):
        os.makedirs(dest_dir)
        print(f"Created destination directory: '{dest_dir}'")

    all_files = [f for f in os.listdir(source_dir) if os.path.isfile(os.path.join(source_dir, f))]

    if len(all_files) < num_files:
        print(f"Warning: Only {len(all_files)} files available in source directory. Copying all of them.")
        files_to_copy = all_files
    else:
        files_to_copy = random.sample(all_files, num_files)

    for file_name in files_to_copy:
        source_path = os.path.join(source_dir, file_name)
        dest_path = os.path.join(dest_dir, file_name)
        try:
            shutil.copy(source_path, dest_path)
            print(f"Copied: '{file_name}' to '{dest_dir}'")
        except Exception as e:
            print(f"Error copying '{file_name}': {e}")
```

```python
#courtesy of chatgpt
def copy_images_from_df(df, column_name, source_dir, target_dir, 
                        extensions=None, case_insensitive=False):
    """
    Copy images from source_dir to target_dir if their base filenames 
    appear in df[column_name]. Extensions can be provided.
    
    Args:
        df (pd.DataFrame): DataFrame containing filenames (without extension).
        column_name (str): Name of the column with base filenames.
        source_dir (str): Path to the folder containing images.
        target_dir (str): Path to the folder where matched images will be copied.
        extensions (list[str] or None): List of file extensions to check (e.g. ['.jpg', '.png']).
                                        If None, matches any extension found in source_dir.
        case_insensitive (bool): Whether to match filenames ignoring case. Default = False.
    """
    os.makedirs(target_dir, exist_ok=True)

    # Normalize names
    base_names = set(df[column_name].astype(str))
    #print(base_names)
    if case_insensitive:
        base_names = {name.lower() for name in base_names}

    for filename in os.listdir(source_dir):
        base, ext = os.path.splitext(filename)

        # Skip if extension not in allowed set (if specified)
        #print(ext)
        #print(extensions)
        if extensions and ext.lower() not in [e.lower() for e in extensions]:
            pass
        print('test')
        compare_name = base.lower() if case_insensitive else base
        print(compare_name)
        if compare_name in base_names:
            print('test')
            shutil.copy(
                os.path.join(source_dir, filename),
                os.path.join(target_dir, filename)
            )
        else:
            print('fail')
```
