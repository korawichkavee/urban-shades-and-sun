```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_cities = pd.read_csv('/home/kieran/Documents/Datasets/Global streetscapes/global-streetscapes/code/raw_download/data/worldcities.csv')
```

```python
df_cities['log_pop'] = df_cities['population'].apply(lambda x: np.log(x) if x>0 else np.nan)
df_cities['population'].plot.hist(bins = 100)
#plot cut at 200k?
cut_loc = 200000
plt.axvline(x=cut_loc, color='r', linestyle='--', label='Vertical cut')
```

```python
df_cities['log_pop'].plot.hist(bins = 100)
plt.axvline(x=np.log(cut_loc), color='r', linestyle='--', label='Vertical cut')
```
